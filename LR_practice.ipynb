```python
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

rng = np.random.default_rng(13)
X = rng.normal(10, 2, (2000, 5))
#y = rng.normal(size=2000)
#in the case of non real world data we give a fake weights, bias, and noise, this is used to calculate y
w_true = np.array([5, 4.2, -1.5, 0.05, -5.3])
b_true = 3
noise = rng.normal(scale=1, size=2000)
y = X @ w_true + b_true + noise
```

```python
X_train = X[0:1500]
y_train = y[0:1500]
X_val = X[1500:1750]
y_val = y[1500:1750]
X_test = X[1750:2000]
y_test = y[1750:2000]
```

```python
w = np.zeros(5)
b = 1
lr = 0.01
steps = 100
for i in range(steps):
  #forward pass
  y_hat = X_train @ w + b
  err = y_train - y_hat
  loss = np.mean(err**2)

  y_val_hat = X_val @ w + b
  val_err = y_val - y_val_hat
  val_loss = np.mean(val_err**2)
  if i % 10 == 0:
    print(f'train loss @ {i} = {loss}')
    print(f'val loss @ {i} = {val_loss}')
  grad_w = (2 / len(X_train)) * X_train.T @ err
  grad_b = (2 / len(X_train)) * np.sum(err)#((2 * err) / len(X_train))

  w = w - (lr * grad_w)
  b = b - (lr * grad_b)





```

```python
y_test_hat = X_test @ w + b
err = y_test - y_test_hat
loss = np.mean(err**2)
print(f'test loss = {loss}')
```
