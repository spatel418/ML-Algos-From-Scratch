```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


rng = np.random.default_rng(13)
X = rng.normal(10, 2, (2000, 6))
w_true = np.array([1.5,0.03, 8.5, -4, -1.2, -5.6])
b_true = 2.4
noise = rng.normal(scale=1, size=2000)

z = X @ w_true + b_true
y = z + noise

def sigmoid(logit):
  return 1 / (1 + np.exp(-logit))

y = sigmoid(y)
print(f'y = {y}')
y = y >= 0.5
print(f'y = {y}')

```

```python
X_train = X[:1500]
y_train = y[:1500]
X_val = X[1500:1750]
y_val = y[1500:1750]
X_test = X[1750:2000]
y_test = y[1750:2000]
```

```python
steps = 500
lr = 0.01
b = 1
w = np.zeros(X.shape[1])
for i in range(steps):
  y_hat_train = X_train @ w + b
  y_hat_val = X_val @ w + b
  p = sigmoid(y_hat_train)
  p_val = sigmoid(y_hat_val)
  #err = y_hat_train_sig - y_train
  #MCE
  #loss = -np.mean(np.sum(y * np.log(p), axis=1))
  loss = -np.mean(y_train*np.log(p) + (1-y_train) * np.log(1-p))
  val_loss = -np.mean(y_val*np.log(p_val) + (1-y_val) * np.log(1-p_val))
  if i % 10 == 0:
    print(f'training loss = {loss}')
    print(f'val loss = {val_loss}')

  dL_dp = ((-1 * (y_train / p)) + ((1-y_train) / (1-p))) / len(y_train)
  print(f'dL_dp shape = {dL_dp.shape} | res = {dL_dp}')
  dp_dz = ((np.exp(-y_hat_train)) / ((1 + np.exp(-y_hat_train)) ** 2)) * dL_dp
  grad_w = X_train.T @ dp_dz
  grad_b = np.sum(dp_dz)
  w = w - (lr * grad_w)
  b = b - (lr * grad_b)


  """
  bce = -log(y_hat_train_sig)

  -1/n y * log(p) - (1-y) * log(1-p)

  total = 0

  for c in classes:
    total += y[c] * log(c)

  -1/n * total
  """
```
